## 4. Distribution of Animal Types. I performed a frequency count on the "Animal Type" column to see the total distribution.

```python
# Question 4: Distribution of Animal Types
animal_distribution = df['animal_type'].value_counts()
print("Animal Type Distribution:")
print(animal_distribution)
```

## 5. Adoption Rates by Top 5 Breeds. I found the top 5 breeds by count, then calculated the percentage of "Adoption" outcomes specifically for those five breeds.

```python
# Question 5: Adoption Rates for Top 5 Breeds
top_5_breeds = df[df['animal_type'] == 'Dog']['breed'].value_counts().head(5).index
breed_stats = df[df['breed'].isin(top_5_breeds)]

adoption_by_breed = breed_stats.groupby('breed')['outcome_type'].apply(lambda x: (x == 'Adoption').mean())
print("Adoption Rates for Top 5 Dog Breeds:")
print(adoption_by_breed.sort_values(ascending=False))
```

## 6. Adoption Rates by Top 5 Colorings. Similar to the breed analysis, I identified the top 5 most common colors and calculated the adoption percentage for each.

```python
# Question 6: Adoption Rates for Top 5 Colors
top_5_colors = df['color'].value_counts().head(5).index
color_stats = df[df['color'].isin(top_5_colors)]

adoption_by_color = color_stats.groupby('color')['outcome_type'].apply(lambda x: (x == 'Adoption').mean())
print("Adoption Rates for Top 5 Colors:")
print(adoption_by_color.sort_values(ascending=False))
```

## 7. Spay/Neuter Monthly Estimates. I filtered for animals listed as "Intact Male" or "Intact Female" and averaged these counts across the available months.

```python
# Question 7: Monthly Spay/Neuter Estimates
intact_animals = df[df['sex_upon_outcome'].str.contains('Intact', na=False)]
monthly_surgery_avg = intact_animals.groupby(df['datetime'].dt.to_period('M')).size().mean()

print(f"Estimated animals to be spayed/neutered per month: {monthly_surgery_avg:.2f}")
```
